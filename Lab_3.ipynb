Mathematical Operations on 2D Arrays

```python
import numpy as np

A = np.array([
    [10, 20, 30],
    [40, 50, 60]
])

B = np.array([
    [1, 2, 3],
    [4, 5, 6]
])

print("\nArray A:")
print(A)

print("\nArray B:")
print(B)

# Addition
print("\nAddition (A + B):")
print(A + B)

# Subtraction
print("\nSubtraction (A - B):")
print(A - B)

# Element-wise Multiplication
print("\nElement-wise Multiplication (A * B):")
print(A * B)

# Element-wise Division
print("\nElement-wise Division (A / B):")
print(A / B)

# Matrix Multiplication
C = np.array([
    [1, 2],
    [3, 4],
    [5, 6]
])

print("\nMatrix C:")
print(C)

print("\nMatrix Multiplication (A @ C):")
print(A @ C)
```

Creating and Checking a 3D NumPy Array

```python
import numpy as np

A = np.array([
    [[1, 2], [3, 4]],
    [[5, 6], [7, 8]]
])

print(A)

print("Shape:", A.shape)
print("Dimensions:", A.ndim)
```
