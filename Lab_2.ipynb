Creating a One-Dimensional NumPy Array

```python
import numpy as np

marks = np.array([85, 90, 78, 88, 92])

print(marks)
```

Creating a Two-Dimensional NumPy Array

```python
import numpy as np

student_marks = np.array([
    [85, 90, 78],
    [88, 76, 91],
    [92, 84, 80]
])

print(student_marks)
```

Basic NumPy Array Operations

```python
import numpy as np

arr = np.array([10, 20, 30, 40, 50])

# Display the array
print("Array :", arr)

# Basic operations
print("Maximum Value :", arr.max())
print("Minimum Value :", arr.min())
print("Sum :", arr.sum())
print("Average :", arr.mean())
print("Number of Elements :", arr.size)
```

Array Indexing, Slicing, Arithmetic & Reshaping

```python
import numpy as np

arr = np.array([10, 20, 30, 40, 50])  # Create an array

# Indexing
print("First Element :", arr[0])
print("Last Element :", arr[-1])

# Slicing
print("Elements from Index 1 to 3 :", arr[1:4])

# Arithmetic operations
print("Array + 10 :", arr + 10)
print("Array × 2 :", arr * 2)

# Reshape
new_arr = arr.reshape(5, 1)

print("Reshaped Array:")
print(new_arr)
```

Creating a pandas DataFrame

```python
import pandas as pd

data = {
    "Roll No": [101, 102, 103, 104],
    "Name": ["Ravi", "Priya", "Arun", "Meena"],
    "Marks": [85, 92, 78, 88]
}

df = pd.DataFrame(data)

print(df)
```

DataFrame Operations

```python
import pandas as pd

# Create a DataFrame
data = {
    "Roll No": [101, 102, 103, 104],
    "Name": ["Ravi", "Priya", "Arun", "Meena"],
    "Marks": [85, 92, 78, 88]
}

df = pd.DataFrame(data)

# Display the DataFrame
print("DataFrame")
print(df)

# Display first two rows
print("\nFirst Two Rows")
print(df.head(2))

# Display last two rows
print("\nLast Two Rows")
print(df.tail(2))

# Display column names
print("\nColumn Names")
print(df.columns)

# Display DataFrame shape
print("\nShape")
print(df.shape)

# Display summary information
print("\nInformation")
print(df.info())

# Display statistical summary
print("\nStatistical Summary")
print(df.describe())
```
